Theoretical Questions & Answers

1.What does R-squared represent in a regression model?
- The proportion of variance in the dependent variable explained by the independent variables (model fit quality).

2.What are the assumptions of linear regression?
- Linearity, independence of errors, homoscedasticity, normality of residuals, no multicollinearity, and correct model specification.

3.Difference between R-squared and Adjusted R-squared?
- Adjusted R² penalizes adding irrelevant predictors and adjusts for number of features, preventing artificial inflation of fit.

4.Why use Mean Squared Error (MSE)?
- It measures average squared difference between predictions and actual values, emphasizing larger errors.

5.What does an Adjusted R-squared of 0.85 indicate?
- 85% of variance is explained by the model after adjusting for number of predictors.

6.How check normality of residuals in linear regression?
- Use Q-Q plot, histogram of residuals, or statistical tests (e.g. Shapiro–Wilk).

7.What is multicollinearity, and how does it impact regression?
- High correlation among predictors inflates variance of coefficient estimates, making them unstable.

8.What is Mean Absolute Error (MAE)?
- The average of absolute differences between predicted and actual values (less sensitive to outliers).

9.Benefits of using an ML pipeline?
- Encapsulates preprocessing, feature selection, scaling, and modeling steps for reproducibility and clean workflow.

10.Why is RMSE considered more interpretable than MSE?
- RMSE is in same units as target, making error magnitude directly interpretable.

11.What is pickling in Python, and how is it useful in ML?
- Serialization of Python objects to disk, enabling model saving/loading for reuse.

12.What does a high R-squared value mean?
- Model explains a large portion of variance, but doesn’t guarantee causality or proper specification.

13.What happens if linear regression assumptions are violated?
-  Estimates become biased or inefficient; inference (p-values, confidence intervals) may be invalid.

14.How address multicollinearity in regression?
- Remove or combine correlated features, apply dimensionality reduction (PCA), or use regularization (Ridge/Lasso).

15.How can feature selection improve model performance in regression analysis?
- Removes irrelevant/noisy variables, reducing overfitting and improving generalization.

16.How is Adjusted R-squared calculated?
- $\displaystyle 1 - (1 - R^2) \frac{n - 1}{n - p - 1}$, where $n$=samples, $p$=predictors.

17Why is MSE sensitive to outliers?
- Squaring amplifies large errors disproportionately.

18.Role of homoscedasticity in linear regression?
- Constant variance of errors ensures unbiased, efficient estimates and valid inference.

19.What is RMSE?
- Square root of MSE; measures standard deviation of residuals.

20.Why is pickling considered risky?
- Untrusted pickle files can execute arbitrary code when loaded (security risk).

21.Alternatives to pickling for saving ML models?
- Joblib, ONNX, joblib for large arrays, or framework-specific formats (e.g., TensorFlow SavedModel).

22.What is heteroscedasticity, and why is it a problem?
- Non-constant error variance leading to inefficient estimates and invalid inference.

23.How can interaction terms enhance a regression model's predictive power?
- Capture combined effects of predictors, modeling non-additive relationships.

```python
# 1. Visualize residuals distribution for multiple linear regression on Seaborn diamonds dataset
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

df = sns.load_dataset('diamonds')
features = ['carat','depth','table','x','y','z']
X = df[features]
y = df['price']
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
model = LinearRegression().fit(X_train, y_train)
residuals = y_test - model.predict(X_test)

sns.histplot(residuals, kde=True)
plt.title('Residuals Distribution')
plt.xlabel('Residual')
plt.show()
```

```python
# 2. Calculate and print MSE, MAE, RMSE for a linear regression model
from sklearn.metrics import mean_squared_error, mean_absolute_error
# reuse model and data from (1)
mse = mean_squared_error(y_test, model.predict(X_test))
mae = mean_absolute_error(y_test, model.predict(X_test))
rmse = mse ** 0.5
print(f"MSE: {mse:.2f}, MAE: {mae:.2f}, RMSE: {rmse:.2f}")
```

```python
# 3. Check regression assumptions: scatter for linearity, residuals vs fitted, and VIF for multicollinearity\import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Linearity scatter
plt.scatter(model.predict(X_test), y_test)
plt.xlabel('Fitted values'); plt.ylabel('Actual')
plt.show()
# Residuals vs fitted
plt.scatter(model.predict(X_test), residuals)
plt.xlabel('Fitted'); plt.ylabel('Residuals')
plt.axhline(0, color='black'); plt.show()
# VIF
df_vif = X.assign(const=1)
vif = {col: variance_inflation_factor(df_vif.values, i)
       for i, col in enumerate(df_vif.columns) if col!='const'}
print('VIF:', vif)
```

```python
# 4. Pipeline with scaling and evaluate multiple regression models
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import Ridge, Lasso
from sklearn.model_selection import cross_val_score

pipeline = Pipeline([('scale', StandardScaler()), ('lr', LinearRegression())])
scores = cross_val_score(pipeline, X, y, cv=5, scoring='r2')
print('Linear Regression R2:', scores.mean())

ridge_pipe = Pipeline([('scale', StandardScaler()), ('ridge', Ridge())])
print('Ridge R2:', cross_val_score(ridge_pipe, X, y, cv=5, scoring='r2').mean())
```

```python
# 11. Fit polynomial regression models of different degrees on synthetic data and compare R-squared
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# Generate synthetic data
X = np.linspace(0, 10, 100).reshape(-1, 1)
y = 1 + 2 * X.flatten() + 3 * X.flatten()**2 + np.random.randn(100) * 10

for degree in [1, 2, 3, 4, 5]:
    poly = PolynomialFeatures(degree)
    X_poly = poly.fit_transform(X)
    model = LinearRegression().fit(X_poly, y)
    r2 = model.score(X_poly, y)
    print(f"Degree {degree}: R2 = {r2:.3f}")
```

```python
# 14. Check multicollinearity using Variance Inflation Factor (VIF)
import numpy as np
import pandas as pd
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Synthetic dataset with multicollinearity
df = pd.DataFrame({'x1': np.random.rand(100)})
df['x2'] = df['x1'] * 0.9 + np.random.rand(100) * 0.1
df['x3'] = np.random.rand(100)

# Calculate VIF
df_vif = df.copy()
df_vif['const'] = 1
vif = {
    col: variance_inflation_factor(df_vif.values, i)
    for i, col in enumerate(df_vif.columns) if col != 'const'
}
print("VIF:", vif)
```

```python
# 22. Linear regression with categorical features using one-hot encoding on seaborn 'tips' dataset
import seaborn as sns
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Load data
df = sns.load_dataset('tips')

# Specify features and target
categorical = ['sex', 'smoker', 'day', 'time']
numeric = ['total_bill', 'size']
y = df['tip']
X = df[categorical + numeric]

# Preprocessing pipeline
preprocessor = ColumnTransformer([
    ('cat', OneHotEncoder(drop='first'), categorical)
], remainder='passthrough')

pipeline = Pipeline([
    ('pre', preprocessor),
    ('lr', LinearRegression())
])

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

# Fit and evaluate
pipeline.fit(X_train, y_train)
print("R-squared:", pipeline.score(X_test, y_test))
```

```python
# 23. Compare Ridge Regression with Linear Regression on a synthetic dataset and print coefficients and R-squared
import numpy as np
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.model_selection import train_test_split

# Synthetic data
dim = 50
X = np.random.randn(100, dim)
coef_true = np.random.randn(dim)
y = X.dot(coef_true) + np.random.randn(100) * 0.5

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

# Linear Regression
lr = LinearRegression().fit(X_train, y_train)
print("Linear Regression R2:", lr.score(X_test, y_test))
print("LR Coefficients:", lr.coef_)

# Ridge Regression
ridge = Ridge(alpha=1.0).fit(X_train, y_train)
print("Ridge Regression R2:", ridge.score(X_test, y_test))
print("Ridge Coefficients:", ridge.coef_)
```
